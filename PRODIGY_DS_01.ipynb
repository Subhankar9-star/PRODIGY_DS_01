```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load the data (Colab lets you use URLs directly)
url = 'https://raw.githubusercontent.com/datasets/population/master/data/population.csv'
df = pd.read_csv(url)

# 2. Filter for the year 2018 (or any year available)
data_2018 = df[df['Year'] == 2018].sort_values(by='Value', ascending=False).head(10)

# 3. Create the Bar Chart
plt.figure(figsize=(12, 6))
sns.barplot(x='Value', y='Country Name', data=data_2018, palette='magma')

plt.title('Top 10 Most Populous Countries (2018)')
plt.xlabel('Total Population')
plt.ylabel('Country')
plt.show()
```
